```python
# -*- coding: utf-8 -*-
# -*- coding: utf-8 -*-
from __future__ import division
import pandas as pd
import math
import random
import string
import pickle
from sklearn.preprocessing import StandardScaler
import digitalization
import numpy as np
import BPNN

BPNN.factor(1000, 400) # You can adjust the training steps and how many data you want to used to train the model.

df=pd.read_csv('data_to_predict.csv')
input_data = df.drop(['ENGG2020'],axis=1)
digitalization.predict(input_data)
sid = pd.read_csv('predict_data.csv',usecols=['Student_ID'])
data = pd.read_csv('predict_data.csv',usecols=['Admission_Channel','Secondary_School','Exam','Math_Background','MATH1510','ENGG1130'])
pd.DataFrame(sid).to_csv('sid.csv',index=None)
sc = StandardScaler()
data = sc.fit_transform(data)
pd.DataFrame(data).to_csv('feature.csv',index=None)
f1 = pd.read_csv('sid.csv')
f2 = pd.read_csv('feature.csv')
file = [f1,f2]
train1 = pd.concat(file,axis=1)
train1.to_csv("temporary" + ".csv", index=0, sep=',')

level = {0:'A',
                 1:'B-',
                 2:'B',
                 3:'B+',
                 4:'C',
                 5:'D',
                 6:'F'}

def sigmoid(x):
    return 1/(1+math.exp(-x))
    # return math.tanh(x)

def dsigmoid(y):
    return y - y**2
    # return 1 - y**2

class NN:
    def __init__(self, ni, nh, no):
        self.ni = ni + 1
        self.nh = nh
        self.no = no

        self.ai = [1.0]*self.ni
        self.ah = [1.0]*self.nh
        self.ao = [1.0]*self.no

        with open('wi.txt', 'rb') as wif:
          self.wi = pickle.load(wif)
        with open('wo.txt', 'rb') as wof:
          self.wo = pickle.load(wof)
        with open('ci.txt', 'rb') as cif:
          self.ci = pickle.load(cif)
        with open('co.txt', 'rb') as cof:
          self.co = pickle.load(cof)

    def update(self, inputs):
        # print(inputs)
        if len(inputs) != self.ni-1:
            raise ValueError('Different from number of nodes in input layer！')

        for i in range(self.ni-1):
            self.ai[i] = sigmoid(inputs[i])

        for j in range(self.nh):
            sum = 0.0
            for i in range(self.ni):
                sum = sum + self.ai[i] * self.wi[i][j]
            self.ah[j] = sigmoid(sum)

        for k in range(self.no):
            sum = 0.0
            for j in range(self.nh):
                sum = sum + self.ah[j] * self.wo[j][k]
            self.ao[k] = sigmoid(sum)

        # print self.ao
        return self.ao[:]

    def test(self, patterns):
        count = 0
        grade = []
        for p in patterns:
            # target = level[(p[1].index(1))]
            # print(p[0])
            result = self.update(p[0])
            # print(result)
            index = result.index(max(result))
            # print(level[index])
            grade.append(level[index])
            # count += (target == level[index])
        # accuracy = float(count/len(patterns))
        # print(grade)
        print(len(patterns))

        bpnn_df=pd.DataFrame(grade)
        bpnn_df.columns = ['ENGG2020']
        pd.DataFrame(bpnn_df).to_csv('BPNN_pred_level.csv',index=None)
        f1 = pd.read_csv('predict_data.csv')
        f2 = pd.read_csv('BPNN_pred_level.csv')
        file = [f1,f2]
        train = pd.concat(file,axis=1)
        train.to_csv("BPNN_predict_complete" + ".csv", index=0, sep=',')

nn=NN(6,9,7)

# features 0-3
# labels 4
def factor():
    data = []
    # read dataset
    raw = pd.read_csv('temporary.csv')
    raw_data = raw.values
    raw_feature = raw_data[:,1:7]
    for i in range(len(raw_feature)):
        ele = []
        ele.append(list(raw_feature[i]))
        data.append(ele)
    test = data[0:]
    nn.test(test)

if __name__ == '__main__':
  factor()
```
