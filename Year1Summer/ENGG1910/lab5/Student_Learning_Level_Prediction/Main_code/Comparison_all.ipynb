Comparison of running time.

```python
import KNN_Comparison
import SVC_Comparison
import NB_Comparison
import BPNN_Comparison
import matplotlib.pyplot as plt
import pandas as pd
import time

dataset = pd.read_csv('train_data.csv')
sample_number=len(dataset)
a = []
b = []
c = []
d = []
e = []
f = []
g = []
h = []
for i in range(10,sample_number,5): # Amount of dataset from 10 to all, 5 more data added after each iteration.
  start1=time.time()
  a.append(i)
  KNN_Comparison.KNN_comparison(train_data=i,test_number=0.2,k=3) # You can modify the parameters of the model inside the bracket.
  end1=time.time()
  b.append(end1-start1)
for i in range(10,sample_number,5):
  start2=time.time()
  c.append(i)
  SVC_Comparison.SVC_Linear_comparison(train_data=i,test_number=0.2) # You can modify the parameters of the model inside the bracket.
  end2=time.time()
  d.append(end2-start2)
for i in range(10,sample_number,5):
  start3=time.time()
  e.append(i)
  NB_Comparison.NB_comparison(train_data=i,test_number=0.2) # You can modify the parameters of the model inside the bracket.
  end3=time.time()
  f.append(end3-start3)
for i in range(10,sample_number,5):
  print("progress",i/sample_number) # You can see the progress when use this line of code.
  start4=time.time()
  g.append(i)
  test_number = int(i*0.2) # You can adjust the ratio of test data here.
  BPNN_Comparison.factor(iterations=100, total_data=i, training_data=test_number) # You can adjust the iterations here.
  end4=time.time()
  h.append(end4-start4)
plt.plot(a, b)
plt.plot(c, d)
plt.plot(e, f)
plt.plot(g, h)
plt.legend(['KNN', 'SVC', 'NB', "BPNN(100 iterations)"], loc='upper left') # Don't forget to change the name of the legend.
plt.xlabel("Training data")
plt.ylabel("Running time(s)")
plt.title("Running time comparison")
plt.show()
```
